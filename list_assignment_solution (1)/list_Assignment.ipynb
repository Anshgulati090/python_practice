```python
### Assignment 9: Matrix Transposition

#Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.
def transposed_mat(matrix):
    transposed = [[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))]
    return transposed
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
transposed = transposed_mat(matrix)
print("Original matrix")
for row in matrix:
    print(row)
print("Trasposed")
for row in transposed:
    print(row)
```

```python
def flattend_lst(nested_lst):
    flat_lst = [item for sublist in nested_list for item in sublist]
    return flat_lst


nested_list = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
flat = flattend_lst(nested_lst)
print(flat)

```

```python
def flatened_list(nested_list):
    flat_lst = [item for sub_list in nested_list for item in sub_list]
    return flat_lst
nested_list = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
flat = flatened_list(nested_list)
print("""""""""""""")
print(flat)
print("""""""""""""")
```

```python
def reversed_lst(lst):
    return lst[::-1]
original_lst = [1,2,3,4,5,6]
reversed_list = reversed_lst(original_lst)
print(original_lst)
print(reversed_list)
```

```python
def rotated_lst(lst,n):
    return lst[n:] + lst[:n]
original_lst = [1,2,3,4,5]
rotated_list = rotated_lst(original_lst,4)
print(rotated_list)
```

```python
def list_intersection(lst,lst2):
    return [x for x in lst if x in lst2]

lst1 = [1,2,3,4,5,6]
lst2 = [7,8,9,1,0,6]
intersect = list_intersection(lst1,lst2)
print(intersect)
```
