```python
set1 = set(range(1,11))
set1.add(11)
set1.remove(1)
print(set1)
```

```python
set1 = {1,2,3,4,5}
set2 = set(range(2,11,2))
print(set2)
print(set1.union(set2))
print(set1.intersection(set2))
print(set1.difference(set2))
print(set1.symmetric_difference(set2))
```

```python
sq = set(x**2 for x in range(1,11))
print(sq)
```

```python
sq = set(x**2 for x in range(1,11))
print(sorted(sq))
```

```python
even = set(x for x in range(1,11) if x%2==0)
print(sorted(even))
```

```python
set1 = {1,2,3,4,5,6,8,8,9,4,5}
unique_set = set(set1)
print(set1)
```

```python
set_5 = set(range(1,6))
set_3 = set(range(1,4))
print(set_3.issubset(set_5))
print(set_5.issuperset(set_3))
```

```python
f_set = frozenset(range(1,6))
print(f_set)
```

```python
seti = set(range(1,6))
lst = list(seti)
lst.append(6)
seti = set(lst)
print(seti)
```
